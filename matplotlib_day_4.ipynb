```python
import matplotlib.pyplot as plt
import numpy as np

# Indian Flag
plt.barh(3, 10, color="orange")
plt.barh(2, 10, color="white")
plt.barh(1, 10, color="green")

# Ashoka Chakra
circle = plt.Circle((5, 2), 0.5, fill=False, color="blue", linewidth=2)
plt.gca().add_patch(circle)

# 24 spokes
for angle in np.linspace(0, 2*np.pi, 24, endpoint=False):
    x = [5, 5 + 0.5*np.cos(angle)]
    y = [2, 2 + 0.5*np.sin(angle)]
    plt.plot(x, y, color="blue")

plt.xlim(0, 10)
plt.ylim(0.5, 3.5)
plt.axis("off")

plt.show()
```
