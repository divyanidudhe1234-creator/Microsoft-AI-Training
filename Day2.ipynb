```python
from collections import Counter
a=[1,2,3,5,2,4,6,5,4,3,2,1,5,4,6,7,8,2,1]

frequencies = Counter(a)

print("Frequency count:")
for item, count in frequencies.items():
  print(f"{item}: {count}")
```
